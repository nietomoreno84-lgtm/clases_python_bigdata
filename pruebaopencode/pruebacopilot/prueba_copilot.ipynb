# Lista de empleados
## Objetivos

1. Crear un DataFrame con pandas importando el fichero JSON con la lista de empleados.
2. Generar un DataFrame con el listado de departamentos y el número de empleados por departamento.

```python
import pandas as pd
import json
from pathlib import Path

path_json = Path('data/empleados.json')
with path_json.open('r', encoding='utf-8') as file:
    empleados = json.load(file)

empleados_df = pd.DataFrame(empleados)

# Mostrar el dataframe completo
empleados_df

# Contar empleados por departamento
departamentos_df = (
    empleados_df.groupby('departamento', sort=True)
    .size()
    .reset_index(name='numero_empleados')
)

departamentos_df
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 4))
sns.barplot(
    data=departamentos_df,
    x='departamento',
    y='numero_empleados',
    palette='viridis'
)
plt.title('Número de empleados por departamento')
plt.xlabel('Departamento')
plt.ylabel('Número de empleados')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
