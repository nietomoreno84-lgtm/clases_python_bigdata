## 1. Linea temporal interactiva (px.line)

**Caso de uso**: Visualizar evolución temporar interactiva pudiendo remuestrear, hacer zoom, etc.

**Data Model**: Un DataFrame con fechas en el Eje de las X y valores numericos en el Eje de las Y

```python
import pandas as pd
import plotly.express as px





# Dataset: Crecimiento de usuarios registrados en una aplicación durante una semana
datos_usuarios = {
    "Fecha": ["2026-07-01", "2026-07-02", "2026-07-03", "2026-07-04", "2026-07-05", "2026-07-06", "2026-07-07"],
    "Usuarios_Registrados": [120, 145, 190, 230, 280, 310, 390]
}

# Dataset: Crecimiento de usuarios registrados en una aplicación durante una semana
datos_usuarios = {
    "Fecha": ["2026-07-01", "2026-07-02", "2026-07-03", "2026-07-04", "2026-07-05", "2026-07-06", "2026-07-07"],
    "Usuarios_Registrados": [120, 145, 190, 230, 280, 310, 390]
}
df_usuarios = pd.DataFrame(datos_usuarios)

fig = px.line(
    df_usuarios,
    x="Fecha",
    y="Usuarios_Registrados",
    title="Crecimiento de Usuarios Registrados (Julio 2026)",
    markers=True # añadir puntos en la uniones   
)
fig.show()
```

## 2. Gráfico de barras interactivo (px.bar)

**Casos de uso**: Comparar métricas categóricas viendo detalles exactos cuando pasamos el ratón por encima. 

**Data Model**: Eje X categorias Y numericos

```python
# Dataset: Ingresos anuales de 4 sucursales de una tienda
datos_sucursales = {
    "Sucursal": ["Madrid", "Barcelona", "Valencia", "Sevilla"],
    "Ingresos_Miles": [450, 380, 210, 190],
    "Empleados": [12, 10, 6, 5]  # Información extra que mostraremos
}





df_sucursales = pd.DataFrame(datos_sucursales)

fig = px.bar(
    df_sucursales,
    x="Sucursal",
    y="Ingresos_Miles",
    hover_data=['Empleados'],
    title="Ingresos por sucursal",
    color='Sucursal'
)

fig.show()
fig.write_html('interactivo.html')
```

## 3. Gráfico de Sectores / Donut Interactivo (`px.pie`)
**Caso de uso:** Visualizar porcentajes parciales sobre un total, pudiendo aislar categorías haciendo clic en la leyenda.
**Data Model:** Categorías y porcentajes/valores proporcionales.

```python
# Dataset: Cuota de mercado estimada de 4 aplicaciones de mensajería en un estudio local
datos_apps = {
    "Aplicacion": ["WhatsApp", "Telegram", "Signal", "Otros"],
    "Porcentaje": [72, 18, 6, 4]
}
df_apps = pd.DataFrame(datos_apps)

fig = px.pie(
   df_apps,
   names="Aplicacion",
   values="Porcentaje",
   title="Cuota de mercado de apps de mensajería",
   hole=0.4
)

fig.show()
```

## 4. Gráfico de Caja Interactivo (`px.box`)
**Caso de uso:** Analizar distribuciones viendo los datos exactos del boxplot (mediana, máximo, mínimo, cuartiles) al colocar el cursor encima.
**Data Model:** Categoría en X, valor continuo en Y.


```python
# Dataset: Tiempo de carga (en segundos) de una web usando diferentes navegadores (Chrome vs Firefox)
datos_web = {
    "Navegador": ["Chrome"]*7 + ["Firefox"]*7,
    "Tiempo_Carga": [
        1.2, 1.5, 1.1, 1.4, 1.8, 1.3, 0.9,  # Chrome
        1.8, 2.2, 1.9, 1.5, 2.4, 2.0, 1.7   # Firefox
    ]
}

df_web = pd.DataFrame(datos_web)

fig = px.box(
    df_web,
    x="Navegador",
    y="Tiempo_Carga",
    color="Navegador",
    points="all", # dibuja todos los puntos de muestra indidividuales al lado de las cajas
    title="Comparativa del tiempo de cargar del los navegadores"
)

fig.show()
```
