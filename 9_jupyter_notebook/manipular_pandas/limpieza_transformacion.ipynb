# Limpieza y transformacion de datos.

Trabajamos con BBDD de empleados que tiene valores nulos. Debes:
    1- Rellenar los salarios faltantes NaN con el salario medio de resto de la plantilla
    2- Eliminar las filas donde el departamento sea nulo.
    3- Crear una nueva columna que se llame Salario_Anual multiplicando el salario mensual por 12

```python
import pandas as pd
import numpy as np

datos_empleados = {
    'Nombre': ['Juan', 'Sara', 'Pedro', 'Laura', 'Mike'],
    'Departamento': ['IT', 'Ventas', np.nan, 'IT', 'Marketing'],
    'Salario_Mensual': [2000, np.nan, 1800, 2500, np.nan]
}

df_empleados = pd.DataFrame(datos_empleados)
df_empleados
```

1- Rellenar los salarios faltantes NaN con el salario medio de resto de la plantilla

```python
media_salarial = df_empleados['Salario_Mensual'].mean()
media_salarial

# para insertar el valor de la media dentro de los campos Salario mensual vacios usamos el metodo fillna()
df_empleados['Salario_Mensual'].fillna(media_salarial)
```
